```python
import numpy as np
```

## 1. Datos

Carga los datos de tiempos de pedido y separa el número de productos (`x`) del tiempo observado (`y`).

```python
data = np.loadtxt(
    "../data/order_times.csv",
    delimiter=",",
    skiprows=1
)

x = data[:, 0]
y = data[:, 1]

print("x:", x)
print("y:", y)
```

## 2. Función de predicción

Completa la función para calcular las predicciones del modelo lineal.

```python
def predict(x, w, b):
    # Completar
    return w * x + b
```

## 3. Función de pérdida

Completa la función para calcular el error cuadrático medio (MSE).

```python
def mse_loss(y_true, y_pred):
    # Completar
    return np.mean((y_true - y_pred) ** 2)
```

## 4. Derivada numérica

Implementa una aproximación numérica de la derivada usando la función `quadratic`.

```python
def quadratic(x):
    return x**2


def numerical_derivative(
    f,
    x,
    h=1e-5
):
    # Completar
    return (f(x + h) - f(x)) / h
```

### 4.1. Prueba de derivada numérica

Usa `numerical_derivative(quadratic, 3)` para aproximar la derivada de $x^2$ en $x=3$.

```python
derivative_at_3 = numerical_derivative(quadratic, 3)

print(
    "Derivada aproximada de x^2 en x=3:",
    derivative_at_3
)
```

## 5. Gradientes de la regresión lineal

Completa la función para calcular los gradientes respecto a `w` y `b`.

```python
def linear_regression_gradients(
    x,
    y,
    w,
    b
):
    # Completar

    n = len(x)
    y_pred = predict(x, w, b)

    dw = (-2/n) * np.sum(x * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    return dw, db
```

## 6. Paso de Gradient Descent

Implementa un paso de actualización de los parámetros usando los gradientes y `learning_rate`.

```python
def gradient_descent_step(
    x,
    y,
    w,
    b,
    learning_rate
):
    # Completar
    dw, db = linear_regression_gradients(x, y, w, b)
    w -= learning_rate * dw
    b -= learning_rate * db

    return w, b
```

## 7. Función de entrenamiento

Entrena la regresión lineal durante el número indicado de épocas.

```python
def train_linear_regression(
    x,
    y,
    learning_rate=0.01,
    epochs=1000
):
    # TODO: entrenar el modelo y reemplazar estos valores

    w = 0.0
    b = 0.0

    for epoch in range(epochs):
        y_pred = predict(x, w, b)
        loss = mse_loss(y, y_pred)

        w, b = gradient_descent_step(x, y, w, b, learning_rate)

        if epoch % 100 == 0:
            print(epoch, loss, w, b)

    return w, b
```

## 8. Modelo inicial

Calcula la predicción y la pérdida iniciales usando los parámetros indicados.

```python
initial_w = 0.0
initial_b = 0.0

# TODO: calcular la predicción inicial usando initial_w e initial_b
initial_prediction = predict(x, initial_w, initial_b)

# TODO: calcular la pérdida inicial
initial_loss = mse_loss(y, initial_prediction)
```

## 9. Entrenamiento del modelo

Entrena el modelo con los datos cargados.

```python
w, b = train_linear_regression(
    x,
    y,
    learning_rate=0.01,
    epochs=1000
)
```

## 10. Modelo final

Calcula las predicciones y la pérdida después del entrenamiento, y muestra los resultados.

```python
# TODO: calcular las predicciones finales
final_prediction = predict(x, w, b)

# TODO: calcular la pérdida final
final_loss = mse_loss(y, final_prediction)

print("Pérdida inicial:", initial_loss)
print("Pérdida final:", final_loss)
print("Valor final de w:", w)
print("Valor final de b:", b)
```

## 11. Nueva predicción

Predice el tiempo para un pedido con 7 productos usando los parámetros entrenados.

```python
x_new = 7

prediction = predict(x_new, w, b)

print(
    "Predicción para un pedido con 7 productos:",
    prediction
)
```
